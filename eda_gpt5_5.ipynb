Overall, step1, step2&3

```python
import pandas as pd
import numpy as np

df = df_55.copy()

df = df.dropna(
  subset=[
    "run2_initial_answer",
    "run2_initial_confidence",
    "run2_answer",
    "run2_confidence",
    "answer_idx",
    "meta_info"
  ]
).copy()

df["run2_initial_answer"] = df["run2_initial_answer"].astype(str).str.strip()
df["run2_answer"] = df["run2_answer"].astype(str).str.strip()
df["answer_idx"] = df["answer_idx"].astype(str).str.strip()

df["run2_initial_confidence"] = pd.to_numeric(
  df["run2_initial_confidence"],
  errors="coerce"
)

df["run2_confidence"] = pd.to_numeric(
  df["run2_confidence"],
  errors="coerce"
)

df = df.dropna(
  subset=[
    "run2_initial_confidence",
    "run2_confidence"
  ]
).copy()

df["meta_subset"] = np.where(
  df["meta_info"].astype(str).str.lower().str.contains("step1"),
  "Step 1",
  "Step 2 & 3"
)

df["initial_correct"] = df["run2_initial_answer"].eq(df["answer_idx"])
df["final_correct"] = df["run2_answer"].eq(df["answer_idx"])


def ece_score(correct, confidence, n_bins=10):
  correct = np.asarray(correct).astype(int)
  confidence = np.asarray(confidence).astype(float)

  bins = np.linspace(0, 1, n_bins + 1)
  ece = 0

  for i in range(n_bins):
    lower = bins[i]
    upper = bins[i + 1]

    if i == 0:
      mask = (confidence >= lower) & (confidence <= upper)
    else:
      mask = (confidence > lower) & (confidence <= upper)

    if mask.sum() > 0:
      bin_acc = correct[mask].mean()
      bin_conf = confidence[mask].mean()
      ece += (mask.sum() / len(confidence)) * abs(bin_acc - bin_conf)

  return ece


def mean_sd_string(series):
  if len(series) == 0:
    return "NA"

  mean_value = series.mean()
  sd_value = series.std(ddof=1)

  if pd.isna(sd_value):
    return f"{mean_value:.2f} ± NA"

  return f"{mean_value:.2f} ± {sd_value:.2f}"


def summarize_basic(data, conf_col, correct_col, label):
  out = []

  subsets = [
    ("Overall", data),
    ("Step 1", data[data["meta_subset"].eq("Step 1")]),
    ("Step 2 & 3", data[data["meta_subset"].eq("Step 2 & 3")])
  ]

  for subset_name, d in subsets:
    if len(d) == 0:
      continue

    correct = d[correct_col].astype(int)
    conf = d[conf_col].astype(float)
    conf_prob = conf / 100

    acc = correct.mean() * 100
    mean_conf = conf.mean()

    d_correct = d[d[correct_col]]
    d_incorrect = d[~d[correct_col]]

    calibration_gap = mean_conf - acc
    ece = ece_score(correct, conf_prob, n_bins=10) * 100
    brier = np.mean((conf_prob - correct) ** 2)

    out.append({
      "Response": label,
      "Subset": subset_name,
      "N": len(d),
      "Accuracy (%)": round(acc, 2),
      "Confidence, mean ± SD (%)": mean_sd_string(conf),
      "Confidence if correct, mean ± SD (%)": mean_sd_string(d_correct[conf_col]),
      "Confidence if incorrect, mean ± SD (%)": mean_sd_string(d_incorrect[conf_col]),
      "Calibration gap (pp)": round(calibration_gap, 2),
      "ECE (pp)": round(ece, 2),
      "Brier score": round(brier, 2)
    })

  return pd.DataFrame(out)


table_initial = summarize_basic(
  data=df,
  conf_col="run2_initial_confidence",
  correct_col="initial_correct",
  label="Baseline"
)

table_final = summarize_basic(
  data=df,
  conf_col="run2_confidence",
  correct_col="final_correct",
  label="Post-reflection"
)

basic_table = pd.concat(
  [
    table_initial,
    table_final
  ],
  ignore_index=True
)

basic_table = basic_table[
  [
    "Response",
    "Subset",
    "N",
    "Accuracy (%)",
    "Confidence, mean ± SD (%)",
    "Confidence if correct, mean ± SD (%)",
    "Confidence if incorrect, mean ± SD (%)",
    "Calibration gap (pp)",
    "ECE (pp)",
    "Brier score"
  ]
]

basic_table
```

```python
import pandas as pd
import numpy as np
from scipy.stats import binom, chi2

df = df_55.copy()

df = df.dropna(
  subset=[
    "run2_initial_answer",
    "run2_initial_confidence",
    "run2_answer",
    "run2_confidence",
    "answer_idx",
    "meta_info"
  ]
).copy()

df["run2_initial_answer"] = df["run2_initial_answer"].astype(str).str.strip()
df["run2_answer"] = df["run2_answer"].astype(str).str.strip()
df["answer_idx"] = df["answer_idx"].astype(str).str.strip()

df["run2_initial_confidence"] = pd.to_numeric(
  df["run2_initial_confidence"],
  errors="coerce"
)

df["run2_confidence"] = pd.to_numeric(
  df["run2_confidence"],
  errors="coerce"
)

df = df.dropna(
  subset=[
    "run2_initial_confidence",
    "run2_confidence"
  ]
).copy()

df["meta_subset"] = np.where(
  df["meta_info"].astype(str).str.lower().str.contains("step1"),
  "Step 1",
  "Step 2 & 3"
)

df["initial_correct"] = df["run2_initial_answer"].eq(df["answer_idx"])
df["final_correct"] = df["run2_answer"].eq(df["answer_idx"])

df["answer_revised"] = df["run2_initial_answer"].ne(df["run2_answer"])

df["correct_to_incorrect"] = (
  df["initial_correct"].eq(True) &
  df["final_correct"].eq(False)
)

df["incorrect_to_correct"] = (
  df["initial_correct"].eq(False) &
  df["final_correct"].eq(True)
)


def mcnemar_pvalue(b, c, continuity_correction=False):
  check_valid = lambda n: isinstance(n, int) or (isinstance(n, float) and n.is_integer())

  if not all(map(check_valid, [b, c])):
    raise ValueError("b and c must be integers.")

  b = int(b)
  c = int(c)

  if b < 0 or c < 0:
    raise ValueError("b and c must be non-negative.")

  total_discordant = b + c

  if total_discordant == 0:
    return np.nan

  n_min, n_max = sorted([b, c])
  corr = int(continuity_correction)

  if total_discordant < 25:
    pvalue = (
      2 * binom.cdf(n_min, total_discordant, 0.5)
      - binom.pmf(n_min, total_discordant, 0.5)
    )
  else:
    chi2_statistic = (abs(n_min - n_max) - corr) ** 2 / total_discordant
    pvalue = chi2.sf(chi2_statistic, 1)

  return pvalue


def summarize_post_reflection(data, model_name="GPT-5.2"):
  out = []

  subsets = [
    ("Overall", data),
    ("Step 1", data[data["meta_subset"].eq("Step 1")]),
    ("Step 2 & 3", data[data["meta_subset"].eq("Step 2 & 3")])
  ]

  for subset_name, d in subsets:
    if len(d) == 0:
      continue

    n = len(d)

    baseline_acc = d["initial_correct"].mean() * 100
    post_acc = d["final_correct"].mean() * 100
    delta_acc = post_acc - baseline_acc

    revision_rate = d["answer_revised"].mean() * 100

    correct_to_incorrect_n = int(d["correct_to_incorrect"].sum())
    incorrect_to_correct_n = int(d["incorrect_to_correct"].sum())

    correct_to_incorrect_pct = correct_to_incorrect_n / n * 100
    incorrect_to_correct_pct = incorrect_to_correct_n / n * 100

    mcnemar_p = mcnemar_pvalue(
      correct_to_incorrect_n,
      incorrect_to_correct_n,
      continuity_correction=False
    )

    out.append({
      "Subset (N)": f"{subset_name} ({n})",
      "Model": model_name,
      "Baseline accuracy (%)": round(baseline_acc, 2),
      "Post-reflection accuracy (%)": round(post_acc, 2),
      "Δ accuracy (pp)": round(delta_acc, 2),
      "Revision rate (%)": round(revision_rate, 2),
      "Correct→Incorrect (%)": round(correct_to_incorrect_pct, 2),
      "Incorrect→Correct (%)": round(incorrect_to_correct_pct, 2),
      "McNemar p-value": round(mcnemar_p, 4) if not pd.isna(mcnemar_p) else np.nan
    })

  return pd.DataFrame(out)


post_reflection_table = summarize_post_reflection(
  data=df,
  model_name="GPT-5.5"
)

post_reflection_table
```
